### Step 1: Import libraries and load data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Load and preview the dataset
df = pd.read_csv("car_purchase_data.csv")
print("Dataset shape:", df.shape)
display(df.head())
print("\nMissing values:\n", df.isnull().sum())
print("\nTarget distribution:\n", df["Purchased"].value_counts())
```

### Step 2: Split features and target

```python
# Use Age and Salary to predict whether a customer purchased a car.
X = df[["Age", "Salary"]]
y = df["Purchased"]

print("Feature columns:", X.columns.tolist())
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

### Step 3: Train/Test Split

```python
# Reserve 25% of the data for testing and use the remaining 75% for training.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=0
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

### Step 4: Feature Scaling

```python
# Fit the scaler only on training data to avoid data leakage.
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaled training data shape:", X_train_scaled.shape)
print("Scaled testing data shape:", X_test_scaled.shape)
```

### Step 5: Train Logistic Regression Model

```python
# Train the logistic regression classifier.
model = LogisticRegression(random_state=0)
model.fit(X_train_scaled, y_train)

print("Model training complete.")
print("Coefficients (Age, Salary):", model.coef_[0])
print("Intercept:", model.intercept_[0])
```

### Step 6: Predict and Evaluate

```python
# Predict the test-set labels and evaluate performance.
y_pred = model.predict(X_test_scaled)
y_prob = model.predict_proba(X_test_scaled)[:, 1]

accuracy = accuracy_score(y_test, y_pred)
print(f"Test accuracy: {accuracy:.2%}")
print("\nConfusion matrix (rows = actual, columns = predicted):")
print(confusion_matrix(y_test, y_pred))
print("\nClassification report:")
print(classification_report(y_test, y_pred, target_names=["Did not purchase", "Purchased"]))

# Show a few predicted probabilities alongside the actual and predicted labels.
results = pd.DataFrame({
    "Age": X_test["Age"].values,
    "Salary": X_test["Salary"].values,
    "Actual Purchased": y_test.values,
    "Predicted Purchased": y_pred,
    "Purchase Probability": y_prob
})
display(results.head(10))
```

### Step 7: Visualization of Decision Regions

```python
# Visualize the model's decision regions in the standardized Age/Salary feature space.
X_plot = X_train_scaled
y_plot = y_train.to_numpy()

x_min, x_max = X_plot[:, 0].min() - 0.5, X_plot[:, 0].max() + 0.5
y_min, y_max = X_plot[:, 1].min() - 0.5, X_plot[:, 1].max() + 0.5
xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 0.02),
    np.arange(y_min, y_max, 0.02)
)
grid = np.c_[xx.ravel(), yy.ravel()]
Z = model.predict(grid).reshape(xx.shape)

plt.figure(figsize=(9, 6))
plt.contourf(xx, yy, Z, alpha=0.25, cmap=plt.cm.coolwarm)
plt.scatter(X_plot[y_plot == 0, 0], X_plot[y_plot == 0, 1],
            label="Did not purchase (0)", edgecolor="black", s=35)
plt.scatter(X_plot[y_plot == 1, 0], X_plot[y_plot == 1, 1],
            label="Purchased (1)", edgecolor="black", s=35)
plt.xlabel("Age (standardized)")
plt.ylabel("Salary (standardized)")
plt.title("Logistic Regression Decision Regions (Training Data)")
plt.legend()
plt.tight_layout()
plt.show()
```

### Reflection Answers

**1. Logistic regression outputs probabilities — why can it still be used for classification?**  
Logistic regression estimates the probability that a sample belongs to the positive class. A decision threshold, commonly 0.5, converts that probability into a class label: probabilities at or above the threshold are classified as 1 (purchased), and probabilities below it are classified as 0 (did not purchase).

**2. Based on your model, which customers are more likely to purchase a car?**  
In this dataset, customers who are older and have higher salaries generally appear more likely to purchase a car. The model combines Age and Salary to estimate each customer's purchase probability. This is a pattern in the data, not a guarantee about any individual customer.

**3. What factors might cause low accuracy in this task?**  
Accuracy may be limited because the model uses only Age and Salary, while purchase decisions can also depend on factors such as expenses, savings, credit history, family needs, and current vehicle ownership. Logistic regression also learns a linear decision boundary in the scaled feature space, so it may not capture more complex patterns. The train/test split can also affect the measured accuracy.

**4. Propose additional features or preprocessing steps that could improve accuracy.**  
Potential improvements include evaluating useful features such as credit score, monthly expenses, employment status, and existing vehicle ownership if those data are available; checking for missing values and outliers; comparing results across several stratified splits or cross-validation; and tuning the regularization strength and classification threshold using training/validation data. Additional features should only be included when they are relevant and reliably collected.
